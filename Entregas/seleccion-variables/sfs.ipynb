## Implementación SFS

```python
import numpy as np
```

```python
def merito(S):
    return int(np.random.rand() * 10 + 1)
```

```python
np.random.seed(42)

F = set(['x1', 'x2','x3','x4','x5','x6','x7','x8','x9'])
S = set([])

k = 6
while (len(S) < 6):

    maximo = -np.inf

    for f in F:
        S_tmp = S | set([f])
        m = merito(S_tmp)

        if m > maximo:
            maximo = m
            f_mejor = f
        print(S_tmp, m)

    print('x_mejor', f_mejor)

    S = S | set([f_mejor])
    F = F - set([f_mejor])
    
    print('S:',S)
    print('F:',F)
    print(10*'-')
```
